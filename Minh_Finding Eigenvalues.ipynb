# Eigendecompositions
## 1. Definition
Let $A \in \mathbb{R}^{n \times n}$ be a square matrix.
A scalar $\lambda \in \mathbb{R}$ and a non-zero vector $\mathbf{v} \in \mathbb{R}^n$ are called an **eigenvalue** and an **eigenvector**, respectively, of the matrix $A$ if they satisfy the equation:

$$A\mathbf{v} = \lambda \mathbf{v}$$

This equation states that applying the matrix $A$ to the vector $\mathbf{v}$ does not change its direction, but only scales it by the factor $\lambda$.


## 2. Finding Eigenvalues

To find the scalar $\lambda$, we rewrite the equation $A\mathbf{v} = \lambda \mathbf{v}$ as follows:

$$A\mathbf{v} - \lambda \mathbf{v} = \mathbf{0}$$
By substituting $\lambda \mathbf{v}$ with $\lambda I \mathbf{v}$, where $I$ is the $n \times n$ identity matrix, we can factor out $\mathbf{v}$: $$(A - \lambda I)\mathbf{v} = \mathbf{0}$$

A non-trivial solution $\mathbf{v} \neq \mathbf{0}$ exists if and only if the matrix $(A - \lambda I)$ is **singular** (non-invertible). This condition is met when its determinant is zero.

Therefore, the eigenvalues $\lambda$ of $A$ are obtained by solving:

$$\det(A - \lambda I) = 0$$

Once the eigenvalues $\lambda_i$ are found, we substitute each $\lambda_i$ back into the original system:

$$(A - \lambda_i I)\mathbf{v} = \mathbf{0}$$

## Example: Calculating Eigenvalues and Eigenvectors

Consider the $2 \times 2$ matrix:
$$A = \begin{pmatrix} 1 & 2 \\ 5 & 4 \end{pmatrix}$$

### 1. Finding Eigenvalues ($\lambda$)
The eigenvalues $\lambda$ are found by solving the equation:
$$\det(A - \lambda I) = 0$$

Given the identity matrix $I = \begin{pmatrix} 1 & 0 \\ 0 & 1 \end{pmatrix}$, we have:
$$A - \lambda I = \begin{pmatrix} 1 & 2 \\ 5 & 4 \end{pmatrix} - \lambda \begin{pmatrix} 1 & 0 \\ 0 & 1 \end{pmatrix} = \begin{pmatrix} 1 - \lambda & 2 \\ 5 & 4 - \lambda \end{pmatrix}$$

Set the determinant equal to zero:
$$\det(A - \lambda I) = \left| \begin{matrix} 1 - \lambda & 2 \\ 5 & 4 - \lambda \end{matrix} \right| = 0$$

Evaluating the determinant of the $2 \times 2$ matrix:
$$(1 - \lambda)(4 - \lambda) - (2)(5) = 0$$
$$(4 - \lambda - 4\lambda + \lambda^2) - 10 = 0$$
$$\lambda^2 - 5\lambda - 6 = 0$$

We solve the quadratic equation $\lambda^2 - 5\lambda - 6 = 0$. Factoring the equation:
$$(\lambda - 6)(\lambda + 1) = 0$$

The eigenvalues are:
$$\lambda_1 = 6 \quad \text{and} \quad \lambda_2 = -1$$

### 2. Finding Eigenvectors ($\mathbf{v}$)

We solve the homogeneous system $(A - \lambda I)\mathbf{v} = \mathbf{0}$ for each eigenvalue.

### Case 1: For $\lambda_1 = 6$

Substitute $\lambda = 6$ into $(A - \lambda I)\mathbf{v} = \mathbf{0}$:
$$\begin{pmatrix} 1 - 6 & 2 \\ 5 & 4 - 6 \end{pmatrix} \begin{pmatrix} x \\ y \end{pmatrix} = \begin{pmatrix} 0 \\ 0 \end{pmatrix}$$

$$\begin{pmatrix} -5 & 2 \\ 5 & -2 \end{pmatrix} \begin{pmatrix} x \\ y \end{pmatrix} = \begin{pmatrix} 0 \\ 0 \end{pmatrix}$$

This matrix equation yields the single independent equation:
$$-5x + 2y = 0 \quad \implies \quad 2y = 5x \quad \implies \quad y = \frac{5}{2}x$$

Choosing a value $x = 2$, we get $y = 5$.
The corresponding eigenvector is:
$$\mathbf{v}_1 = \begin{pmatrix} 2 \\ 5 \end{pmatrix}$$

### Case 2: For $\lambda_2 = -1$

Substitute $\lambda = -1$ into $(A - \lambda I)\mathbf{v} = \mathbf{0}$:
$$\begin{pmatrix} 1 - (-1) & 2 \\ 5 & 4 - (-1) \end{pmatrix} \begin{pmatrix} x \\ y \end{pmatrix} = \begin{pmatrix} 0 \\ 0 \end{pmatrix}$$

$$\begin{pmatrix} 2 & 2 \\ 5 & 5 \end{pmatrix} \begin{pmatrix} x \\ y \end{pmatrix} = \begin{pmatrix} 0 \\ 0 \end{pmatrix}$$

This leads to the single independent equation:
$$2x + 2y = 0 \quad \implies \quad x + y = 0 \quad \implies \quad y = -x$$

Choosing a convenient value $x = 1$, we get $y = -1$.
The corresponding eigenvector is:
$$\mathbf{v}_2 = \begin{pmatrix} 1 \\ -1 \end{pmatrix}$$

## Exercise 1:
**Problem:** What are the eigenvalues and eigenvectors of $A = \begin{pmatrix} 2 & 1 \\ 1 & 2 \end{pmatrix}$ ?

### 1. Finding Eigenvalues ($\lambda$)

We solve the equation $\det(A - \lambda I) = 0$.

$$A - \lambda I = \begin{pmatrix} 2 - \lambda & 1 \\ 1 & 2 - \lambda \end{pmatrix}$$

$$\det(A - \lambda I) = (2 - \lambda)(2 - \lambda) - (1)(1) = 0$$

$$(2 - \lambda)^2 - 1 = 0$$

$$4 - 4\lambda + \lambda^2 - 1 = 0$$

$$\lambda^2 - 4\lambda + 3 = 0$$

$$(\lambda - 3)(\lambda - 1) = 0$$

The eigenvalues are:
$$\lambda_1 = 3 \quad \text{and} \quad \lambda_2 = 1$$

### 2. Finding Eigenvectors ($\mathbf{v}$)

#### Case 1: For $\lambda_1 = 3$
Substitute $\lambda = 3$ into $(A - \lambda I)\mathbf{v} = \mathbf{0}$:
$$\begin{pmatrix} 2 - 3 & 1 \\ 1 & 2 - 3 \end{pmatrix} \begin{pmatrix} x \\ y \end{pmatrix} = \begin{pmatrix} 0 \\ 0 \end{pmatrix}$$

$$\begin{pmatrix} -1 & 1 \\ 1 & -1 \end{pmatrix} \begin{pmatrix} x \\ y \end{pmatrix} = \begin{pmatrix} 0 \\ 0 \end{pmatrix}$$

This implies $-x + y = 0$, or $y = x$.
Choosing $x = 1$, the corresponding eigenvector is:
$$\mathbf{v}_1 = \begin{pmatrix} 1 \\ 1 \end{pmatrix}$$

#### Case 2: For $\lambda_2 = 1$
Substitute $\lambda = 1$ into $(A - \lambda I)\mathbf{v} = \mathbf{0}$:
$$\begin{pmatrix} 2 - 1 & 1 \\ 1 & 2 - 1 \end{pmatrix} \begin{pmatrix} x \\ y \end{pmatrix} = \begin{pmatrix} 0 \\ 0 \end{pmatrix}$$

$$\begin{pmatrix} 1 & 1 \\ 1 & 1 \end{pmatrix} \begin{pmatrix} x \\ y \end{pmatrix} = \begin{pmatrix} 0 \\ 0 \end{pmatrix}$$

This implies $x + y = 0$, or $y = -x$.
Choosing $x = 1$, the corresponding eigenvector is:
$$\mathbf{v}_2 = \begin{pmatrix} 1 \\ -1 \end{pmatrix}$$

**Conclusion:**
The eigenvalues are $\lambda_1=3$ and $\lambda_2=1$. The corresponding eigenvectors are $\mathbf{v}_1=\begin{pmatrix} 1 \\ 1 \end{pmatrix}$ and $\mathbf{v}_2=\begin{pmatrix} 1 \\ -1 \end{pmatrix}$.

## Exercise 2:
**Problem:** Eigenvalues and Eigenvectors of $A = \begin{pmatrix} 2 & 1 \\ 0 & 2 \end{pmatrix}$

### 1. Finding Eigenvalues ($\lambda$)

We solve the equation $\det(A - \lambda I) = 0$.

$$A - \lambda I = \begin{pmatrix} 2 - \lambda & 1 \\ 0 & 2 - \lambda \end{pmatrix}$$

$$\det(A - \lambda I) = (2 - \lambda)(2 - \lambda) - (1)(0) = 0$$

$$(2 - \lambda)^2 = 0$$

The equation gives a repeated root (algebraic multiplicity of 2):
$$\lambda_1 = 2 \quad \text{and} \quad \lambda_2 = 2$$

### 2. Finding Eigenvectors ($\mathbf{v}$)

### Case 1: For $\lambda = 2$ (Repeated Eigenvalue)
Substitute $\lambda = 2$ into $(A - \lambda I)\mathbf{v} = \mathbf{0}$:
$$\begin{pmatrix} 2 - 2 & 1 \\ 0 & 2 - 2 \end{pmatrix} \begin{pmatrix} x \\ y \end{pmatrix} = \begin{pmatrix} 0 \\ 0 \end{pmatrix}$$

$$\begin{pmatrix} 0 & 1 \\ 0 & 0 \end{pmatrix} \begin{pmatrix} x \\ y \end{pmatrix} = \begin{pmatrix} 0 \\ 0 \end{pmatrix}$$

This matrix equation yields the system:
$$0x + 1y = 0 \quad \implies \quad y = 0$$
$$0x + 0y = 0 \quad \text{(always true)}$$

The eigenvector must satisfy $y=0$. $x$ can be any non-zero number.
Choosing $x = 1$, the corresponding eigenvector is:
$$\mathbf{v} = \begin{pmatrix} 1 \\ 0 \end{pmatrix}$$

**What is strange about this example compared to the previous one?**

The matrix $A = \begin{pmatrix} 2 & 1 \\ 0 & 2 \end{pmatrix}$ has a single eigenvalue $\lambda=2$ with an **algebraic multiplicity** of 2. However, the corresponding dimension of the eigenspace is only 1. We found only one eigenvector $\mathbf{v}=\begin{pmatrix} 1 \\ 0 \end{pmatrix}$.

* **Exercise 1:** Algebraic Multiplicity (AM) = Geometric Multiplicity (GM) for both $\lambda$ (AM=GM=1). The matrix is diagonalizable.
* **Exercise 2:** For $\lambda=2$, AM=2 but GM=1. The matrix is **defective** and **not diagonalizable**.

## Exercise 3: Smallest Eigenvalue Estimation

**Problem:** Without computing the eigenvalues, is it possible that the smallest eigenvalue of the following matrix is less than 0.5?
$$A = \begin{pmatrix} 3.0 & 0.1 & 0.3 & 1.0 \\ 0.1 & 1.0 & 0.1 & 0.2 \\ 0.3 & 0.1 & 5.0 & 0.0 \\ 1.0 & 0.2 & 0.0 & 1.8 \end{pmatrix}$$

### 1. Gerschgorin Circle Theorem

This problem can be solved mentally by applying the **Gerschgorin Circle Theorem**. This theorem states that every eigenvalue of a matrix $A$ lies within at least one of the Gerschgorin discs $D_i$.

The Gerschgorin disc $D_i$ for row $i$ is centered at $a_{ii}$ (the diagonal element) and has a radius $R_i$ equal to the sum of the absolute values of the off-diagonal elements in that row:
$$D_i: |\lambda - a_{ii}| \le R_i \quad \text{where} \quad R_i = \sum_{j \neq i} |a_{ij}|$$

### 2. Calculation of Gerschgorin Discs

We calculate the center $a_{ii}$ and radius $R_i$ for each row:

* **Row 1:** $a_{11} = 3.0, $$R_1 = |0.1| + |0.3| + |1.0| = 1.4, $$D_1$: $|\lambda - 3.0| \le 1.4$. (Range: $[1.6, 4.4]$)

* **Row 2:** $a_{22} = 1.0, $$R_2 = |0.1| + |0.1| + |0.2| = 0.4, $$D_2$: $|\lambda - 1.0| \le 0.4$. (Range: $[0.6, 1.4]$)

* **Row 3:** $a_{33} = 5.0, $$R_3 = |0.3| + |0.1| + |0.0| = 0.4, $$D_3$: $|\lambda - 5.0| \le 0.4$. (Range: $[4.6, 5.4]$)

* **Row 4:** $a_{44} = 1.8, $$R_4 = |1.0| + |0.2| + |0.0| = 1.2, $$D_4$: $|\lambda - 1.8| \le 1.2$. (Range: $[0.6, 3.0]$)

The Gerschgorin Theorem guarantees that all eigenvalues $\lambda$ must lie within the union of these intervals:
$$\lambda \in ([1.6, 4.4] \cup [0.6, 1.4] \cup [4.6, 5.4] \cup [0.6, 3.0])$$

The smallest possible value for any eigenvalue is the minimum boundary of all intervals, which is $\min(1.6, 0.6, 4.6, 0.6) = 0.6$.

Since the minimum lower bound for all eigenvalues is $\mathbf{0.6}$, the smallest eigenvalue **cannot** be less than 0.5.
